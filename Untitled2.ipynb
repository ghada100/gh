Assignment 1: Advanced Pixel-Level Manipulation

```python
import cv2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# 1. Loading and Inspecting Images
img1_path = '/content/Gemini_Generated_Image_2dh1zu2dh1zu2dh1.jpg'
img2_path = '/content/images (1).jpg'

img1_bgr = cv2.imread(img1_path)
img2_bgr = cv2.imread(img2_path)

img1 = cv2.cvtColor(img1_bgr, cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(img2_bgr, cv2.COLOR_BGR2RGB)

print(f"Image 1 Shape: {img1.shape}, Type: {img1.dtype}")
print(f"Image 2 Shape: {img2.shape}, Type: {img2.dtype}")

# Display Original Images
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.title("Image 1 (RGB)")
plt.imshow(img1)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Image 2 (RGB)")
plt.imshow(img2)
plt.axis('off')
plt.show()
```

```python
# 2. Crop 30x30 Patch & Paste on Image 2 to create Image3
h1, w1, _ = img1.shape
h2, w2, _ = img2.shape

rand_y1 = np.random.randint(0, h1 - 30)
rand_x1 = np.random.randint(0, w1 - 30)

patch = img1[rand_y1:rand_y1+30, rand_x1:rand_x1+30].copy()

Image3 = img2.copy()
rand_y2 = np.random.randint(0, h2 - 30)
rand_x2 = np.random.randint(0, w2 - 30)

Image3[rand_y2:rand_y2+30, rand_x2:rand_x2+30] = patch

# Display Image3 Result
plt.figure(figsize=(5, 5))
plt.title("Image3 (Pasted Patch)")
plt.imshow(Image3)
plt.axis('off')
plt.show()
```

```python
# 3. Convert Images to Grayscale
gray_img1 = cv2.cvtColor(img1, cv2.COLOR_RGB2GRAY)
gray_img2 = cv2.cvtColor(img2, cv2.COLOR_RGB2GRAY)
gray_Image3 = cv2.cvtColor(Image3, cv2.COLOR_RGB2GRAY)

# Display Grayscale Images
plt.figure(figsize=(10, 4))
plt.subplot(1, 3, 1)
plt.title("Gray Image 1")
plt.imshow(gray_img1, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Gray Image 2")
plt.imshow(gray_img2, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Gray Image 3")
plt.imshow(gray_Image3, cmap='gray')
plt.axis('off')
plt.show()
```

```python
# 4. Apply Edge Sharpening Kernel on Image3
sharpen_kernel = np.array([
    [0, -1,  0],
    [-1, 5, -1],
    [0, -1,  0]
])

sharpened_Image3 = cv2.filter2D(gray_Image3, -1, sharpen_kernel)

# Display Sharpened Result
plt.figure(figsize=(5, 5))
plt.title("Sharpened Image3")
plt.imshow(sharpened_Image3, cmap='gray')
plt.axis('off')
plt.show()
```

Assignment 2: Object Detection & Scene Summary

```python
# 1. Install Ultralytics and import dependencies
!pip install ultralytics

import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

# Load pre-trained YOLO model
model = YOLO('yolov8n.pt')
print("YOLO Model Loaded Successfully!")
```

```python
# 2. Load Image & Run Prediction
img_path = '/content/zidane.jpg'  # مسار الصورة المطلوب

image_bgr = cv2.imread(img_path)
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

# Run YOLO model with confidence thresholding
conf_threshold = 0.4
results = model(image_rgb, conf=conf_threshold)[0]

print(f"Detection completed! Found {len(results.boxes)} objects.")
```

```python
# 3. Plot Bounding Boxes and Labels Manually
annotated_image = image_rgb.copy()
detected_objects = []

for box in results.boxes:
    # Extract bounding box coordinates, confidence, and class
    x1, y1, x2, y2 = map(int, box.xyxy[0])
    conf = float(box.conf[0])
    cls_id = int(box.cls[0])
    class_name = model.names[cls_id]

    detected_objects.append(class_name)

    # Draw bounding box manually
    cv2.rectangle(annotated_image, (x1, y1), (x2, y2), (0, 255, 0), 2)

    # Put label text manually
    label = f"{class_name} {conf:.2f}"
    cv2.putText(annotated_image, label, (x1, max(y1 - 10, 20)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 0), 2)

# Display Annotated Image
plt.figure(figsize=(8, 8))
plt.imshow(annotated_image)
plt.axis('off')
plt.title("Detected Objects (Manual Plotting)")
plt.show()
```

```python
# 4. Generate & Display Natural Language Descriptive Summary
counts = {}
for obj in detected_objects:
    counts[obj] = counts.get(obj, 0) + 1

summary_parts = [f"{count} {obj}{'s' if count > 1 else ''}" for obj, count in counts.items()]

if summary_parts:
    summary_text = "This scene contains " + ", ".join(summary_parts) + "."
else:
    summary_text = "No objects were detected above the confidence threshold."

# Display image with the text summary below it
plt.figure(figsize=(8, 8))
plt.imshow(annotated_image)
plt.axis('off')

# Write descriptive script below image
plt.figtext(0.5, 0.02, f"Description: {summary_text}",
            wrap=True, horizontalalignment='center', fontsize=12,
            bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))

plt.show()
print("Generated Summary Script:", summary_text)
```

Assignment 3: Image & Video Classification Pipeline
This task focuses on utilizing YOLOv8n-cls to perform image classification on the CIFAR-10 dataset and extend prediction capabilities to a video stream (Birds.MP4). The output overlays real-time prediction labels and confidence scores on every processed video frame.

```python
# 1. Install required packages
!pip install ultralytics seaborn scikit-learn -q

import cv2
import torch
import torchvision
import torchvision.transforms as transforms
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO
from ultralytics.utils.downloads import download

# CIFAR-10 Class Names
cifar10_classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']

# Fast Download CIFAR-10 Direct ZIP via Ultralytics (Bypasses Toronto Server)
url = 'https://github.com/ultralytics/assets/releases/download/v0.0.0/cifar10.zip'
download(url, dir='./data')

# Load Dataset via Torchvision from the downloaded directory
transform = transforms.Compose([transforms.ToTensor()])
testset = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)

# Load Pre-trained YOLO Classification Model
model_base = YOLO('yolov8n-cls.pt')
print("✅ Baseline YOLO Model Loaded Successfully!")
```

```python
# 2. Random Sampling (10 Images) & Baseline Inference
np.random.seed(42)
indices = np.random.choice(len(testset), 10, replace=False)

correct = 0
plt.figure(figsize=(15, 6))

for idx, index in enumerate(indices):
    image_tensor, ground_truth_label = testset[index]

    # Convert tensor to numpy image for inference
    img_np = (image_tensor.numpy().transpose((1, 2, 0)) * 255).astype(np.uint8)

    # Run Inference using baseline YOLO
    results = model_base(img_np, verbose=False)[0]
    pred_class_id = results.probs.top1

    # Check Accuracy
    is_correct = (pred_class_id == ground_truth_label)
    if is_correct:
        correct += 1

    # Plotting
    plt.subplot(2, 5, idx + 1)
    plt.imshow(img_np)
    plt.axis('off')
    title_color = 'green' if is_correct else 'red'
    plt.title(f"GT: {cifar10_classes[ground_truth_label]}\nPred: {cifar10_classes[pred_class_id]}", color=title_color, fontsize=9)

plt.tight_layout()
plt.show()

# Report Subset Accuracy
accuracy = (correct / 10) * 100
print(f"\n--- Part 1 Results ---")
print(f"Correct Predictions: {correct}/10")
print(f"Subset Accuracy: {accuracy:.2f}%")
```

```python
# 3. Train YOLO Classification Model on CIFAR-10 Dataset
model_cifar = YOLO('yolov8n-cls.pt')

# Train for 5 epochs on CIFAR-10
model_cifar.train(data='cifar10', epochs=5, imgsz=32, project='cifar10_yolo', name='train_run')
print("Model Training Completed Successfully!")
```

```python
# 4. Generate & Visualize Multi-Class Confusion Matrix
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

# Load trained weights
best_model = YOLO('cifar10_yolo/train_run/weights/best.pt')

y_true = []
y_pred = []

# Collect predictions on test subset for matrix calculation
eval_indices = np.random.choice(len(testset), 200, replace=False)

for index in eval_indices:
    image_tensor, ground_truth_label = testset[index]
    img_np = (image_tensor.numpy().transpose((1, 2, 0)) * 255).astype(np.uint8)

    res = best_model(img_np, verbose=False)[0]
    pred_label = res.probs.top1

    y_true.append(ground_truth_label)
    y_pred.append(pred_label)

# Plot Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=cifar10_classes, yticklabels=cifar10_classes)
plt.title('CIFAR-10 Multi-Class Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()

# Display Full Classification Metric Report
print("\nClassification Report:\n")
print(classification_report(y_true, y_pred, target_names=cifar10_classes))
```

```python
# 5. Fixed Standalone Video Classification Pipeline
import cv2
import torch
import os
from ultralytics import YOLO
from IPython.display import HTML
from base64 import b64encode

# Load Model
video_model = YOLO('yolov8n-cls.pt')

# Paths
input_video_path = '/content/Birds.MP4'
raw_output_path = '/content/raw_classified.mp4'
final_output_path = '/content/classified_output.mp4'

# Open Video Stream
cap = cv2.VideoCapture(input_video_path)
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps    = int(cap.get(cv2.CAP_PROP_FPS))

# Setup Video Writer using standard mp4v
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(raw_output_path, fourcc, fps, (width, height))

print("Processing Video Frames...")

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    frame_count += 1
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # Run YOLO Classification
    results = video_model(frame_rgb, verbose=False)[0]
    pred_class_id = results.probs.top1
    pred_label = results.names[pred_class_id]
    confidence = float(results.probs.top1conf.item())

    # Draw Label & Confidence Box on Frame
    text_to_display = f"Class: {pred_label.upper()} ({confidence*100:.1f}%)"
    cv2.rectangle(frame, (20, 20), (500, 70), (0, 0, 0), -1)
    cv2.putText(frame, text_to_display, (30, 55),
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)

    out.write(frame)

cap.release()
out.release()
print(f"✅ Processed {frame_count} frames. Converting video for browser playback...")

# Convert video using ffmpeg to web-compatible h264 format
os.system(f"ffmpeg -y -i {raw_output_path} -vcodec libx264 {final_output_path} -loglevel quiet")

# Display Processed Video directly in Colab
mp4 = open(final_output_path, 'rb').read()
data_url = "data:video/mp4;base64," + b64encode(mp4).decode()
HTML(f'<video width=600 controls autoplay loop><source src="{data_url}" type="video/mp4"></video>')
```
