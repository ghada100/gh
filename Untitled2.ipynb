Assignment 1: Advanced Pixel-Level Manipulation

```python
import cv2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# 1. Loading and Inspecting Images
img1_path = '/content/Gemini_Generated_Image_2dh1zu2dh1zu2dh1.jpg'
img2_path = '/content/images (1).jpg'

img1_bgr = cv2.imread(img1_path)
img2_bgr = cv2.imread(img2_path)

img1 = cv2.cvtColor(img1_bgr, cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(img2_bgr, cv2.COLOR_BGR2RGB)

print(f"Image 1 Shape: {img1.shape}, Type: {img1.dtype}")
print(f"Image 2 Shape: {img2.shape}, Type: {img2.dtype}")

# Display Original Images
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.title("Image 1 (RGB)")
plt.imshow(img1)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Image 2 (RGB)")
plt.imshow(img2)
plt.axis('off')
plt.show()
```

```python
# 2. Crop 30x30 Patch & Paste on Image 2 to create Image3
h1, w1, _ = img1.shape
h2, w2, _ = img2.shape

rand_y1 = np.random.randint(0, h1 - 30)
rand_x1 = np.random.randint(0, w1 - 30)

patch = img1[rand_y1:rand_y1+30, rand_x1:rand_x1+30].copy()

Image3 = img2.copy()
rand_y2 = np.random.randint(0, h2 - 30)
rand_x2 = np.random.randint(0, w2 - 30)

Image3[rand_y2:rand_y2+30, rand_x2:rand_x2+30] = patch

# Display Image3 Result
plt.figure(figsize=(5, 5))
plt.title("Image3 (Pasted Patch)")
plt.imshow(Image3)
plt.axis('off')
plt.show()
```

```python
# 3. Convert Images to Grayscale
gray_img1 = cv2.cvtColor(img1, cv2.COLOR_RGB2GRAY)
gray_img2 = cv2.cvtColor(img2, cv2.COLOR_RGB2GRAY)
gray_Image3 = cv2.cvtColor(Image3, cv2.COLOR_RGB2GRAY)

# Display Grayscale Images
plt.figure(figsize=(10, 4))
plt.subplot(1, 3, 1)
plt.title("Gray Image 1")
plt.imshow(gray_img1, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Gray Image 2")
plt.imshow(gray_img2, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Gray Image 3")
plt.imshow(gray_Image3, cmap='gray')
plt.axis('off')
plt.show()
```

```python
# 4. Apply Edge Sharpening Kernel on Image3
sharpen_kernel = np.array([
    [0, -1,  0],
    [-1, 5, -1],
    [0, -1,  0]
])

sharpened_Image3 = cv2.filter2D(gray_Image3, -1, sharpen_kernel)

# Display Sharpened Result
plt.figure(figsize=(5, 5))
plt.title("Sharpened Image3")
plt.imshow(sharpened_Image3, cmap='gray')
plt.axis('off')
plt.show()
```

Assignment 2: Object Detection & Scene Summary

```python
# 1. Install Ultralytics and import dependencies
!pip install ultralytics

import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

# Load pre-trained YOLO model
model = YOLO('yolov8n.pt')
print("YOLO Model Loaded Successfully!")
```

```python
# 2. Load Image & Run Prediction
img_path = '/content/zidane.jpg'  # مسار الصورة المطلوب

image_bgr = cv2.imread(img_path)
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

# Run YOLO model with confidence thresholding
conf_threshold = 0.4
results = model(image_rgb, conf=conf_threshold)[0]

print(f"Detection completed! Found {len(results.boxes)} objects.")
```

```python
# 3. Plot Bounding Boxes and Labels Manually
annotated_image = image_rgb.copy()
detected_objects = []

for box in results.boxes:
    # Extract bounding box coordinates, confidence, and class
    x1, y1, x2, y2 = map(int, box.xyxy[0])
    conf = float(box.conf[0])
    cls_id = int(box.cls[0])
    class_name = model.names[cls_id]

    detected_objects.append(class_name)

    # Draw bounding box manually
    cv2.rectangle(annotated_image, (x1, y1), (x2, y2), (0, 255, 0), 2)

    # Put label text manually
    label = f"{class_name} {conf:.2f}"
    cv2.putText(annotated_image, label, (x1, max(y1 - 10, 20)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 0), 2)

# Display Annotated Image
plt.figure(figsize=(8, 8))
plt.imshow(annotated_image)
plt.axis('off')
plt.title("Detected Objects (Manual Plotting)")
plt.show()
```

```python
# 4. Generate & Display Natural Language Descriptive Summary
counts = {}
for obj in detected_objects:
    counts[obj] = counts.get(obj, 0) + 1

summary_parts = [f"{count} {obj}{'s' if count > 1 else ''}" for obj, count in counts.items()]

if summary_parts:
    summary_text = "This scene contains " + ", ".join(summary_parts) + "."
else:
    summary_text = "No objects were detected above the confidence threshold."

# Display image with the text summary below it
plt.figure(figsize=(8, 8))
plt.imshow(annotated_image)
plt.axis('off')

# Write descriptive script below image
plt.figtext(0.5, 0.02, f"Description: {summary_text}",
            wrap=True, horizontalalignment='center', fontsize=12,
            bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))

plt.show()
print("Generated Summary Script:", summary_text)
```
